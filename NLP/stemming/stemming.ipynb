Stemming is the process of reducing a word to its base/root-like form by removing prefixes or suffixes.

The important thing is that the result does not necessarily have to be a real English word.

Simple example
playing   → play
played    → play
plays     → play
player    → player

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
words = ["eating", "eats", "eaten","writing","writes","programming","programs", "programers", "history"]
```

```python
for word in words:
    print(word+ "---->"+stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
stemming.stem('setting')
```

```python
# LAncaster Stemming algorithm
from nltk.stem import LancasterStemmer
```

```python
lancaster  = LancasterStemmer()
for word in words:
    print(word+ "---->"+lancaster.stem(word))
```

RegexpStemmer class
NLTK has RegexpStemmer class with the help of which we can easily implement Rgurlar Expression Stemmer algorithms. it basically takes a single regular expression and remove any prefix or suffix that matches the expression. Let us see an example

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem("eating")
```

```python
reg_stemmer.stem("ingplaying")
```

```python
#Snowballstemmer 

from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english',ignore_stopwords=False)
```

```python
for word in words:
    print(word+ "---->"+snowballstemmer.stem(word))
```

```python
print(snowballstemmer.stem("fairly"))
print(snowballstemmer.stem("sportingly"))
```

Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we well get after lemmatization is called 'lemma' which is a root wordd rather than root stem, the output of stemming, after lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordnetLemmaitzer class which is a thin wrapper around the wordnet corpus, this class uses morphy() function to the Wordnet CorpusReader class to find a lemma. let us understand it with an example - 

```python
import nltk 
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
'''
POS - noun n
verb - v
adjective - a
adverb - a
'''
for word in words:
    print(word+ "---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
print(lemmatizer.lemmatize("got",pos="v"))
```

```python
print(lemmatizer.lemmatize("took",pos='v'))
```
