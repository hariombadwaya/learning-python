Function - function is a block of code that perfroms a specific task , it helps in organizing code 

```python
def even_or_odd(num):
    if num%2==0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")

even_or_odd(5)
```

```python
even_or_odd(4)
```

```python
#function with multiple arguments
def add(num1,num2):
     return num1+num2
   


ans = add(2,90)
print(ans)
```

```python
# we can give default parameter
def greet(name= "Guest"):
    print(f"{name} welcome to the pardise")

greet()
```

```python
greet("hariom")
```

```python
# function with variable parameters
# Positional arguments
def print_num(*args):
    for num in args:
        print(num)

print_num(1,2,3,"hey")
```

```python
#keywords arguments
# keywords arguments accepts arguments in key value pairs
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} : {value}")

print_details(name="ram",age= 23)
```

```python
# returns multiple parameters
def para(a,b):
    return a*b,a

ans =para(3,4)
print(ans)
```
