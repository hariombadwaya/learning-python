Lambda function in python -  it is a anonymous function with lambda keyword  , they can have any number of argument but only one expression , they are commonly used for short operations or as arguments to higher order function 

```python
addition = lambda a,b:a+b
type(addition)
print(addition(5,6))
```

```python
check_even = lambda num: num%2==0
check_even(3)
```

map() function -- the map() function applies a given function to all items in an input list (or any other iterable) and return a map object

```python
def square(num):
    return num**2


numbers = [1,2,3,4,5]
list(map(square,numbers))
```

```python
# lambda fucntion with map
list(map(lambda x:x*2, numbers))
```

```python
# map multiple iterables
numbers1 = [1,2,3]
numbers2 = [4,5,6]
list(map(lambda a,b:a+b , numbers1,numbers2))
```

```python
#convert all the fruits in upper case
fruits = ['apple','banana','mango']
print(list(map(str.upper,fruits)))
```

```python
# how to apply map on dictionary
def get_name(person):
    return person['name']

person= [ {"name": "Radha", "age": 23},
       { "name": "krishna", "age": 25}]

list(map(get_name,person))
```

filter() - filter function return the element passing the certain condition


```python
lst = [1,2,3,4,5,6]
ans = list(filter(lambda num:num%2==0,lst))
print(ans)
```
