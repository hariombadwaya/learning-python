```python
#tuple operations
numbers = (1,2,3,4,5,67)
mixed = ("hey", 33,True,1.3)
con= numbers+mixed
print(con)
```

```python
# if i multiple with number it will replicate 
# if i want to perform multiply on each number then i have to loop over this tuple 
print(con * 3)
```

```python
# tuple methods
num = (11,22,33,11,4,5,3,2)
print(num.count(11))
print(num.index(33))
print(num.index(11))
```

```python
# tuple packing
name = "Hariom"
age = 23
city = "Indore"

person = name, age, city

print(person)
```

```python
a,b,c = person
print(a)
print(b)
print(c)
```

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers
#      * operator allow us to unpack multiple values
print(first)
print(middle)  # it will become list
print(last)
```

Dictionaries in python - Dictionaries stores the data in the form of key value pairs 

```python
# creating empty dictionary
empty = {}
print(empty)
print(type(empty))
```

```python
# another way of creating empty dictionary
dis = dict()
print(dis)
```

```python
#Accessing dictionary elements
student ={"name":"hariom", "age": 23 , "city":"Indore"}
print(student)
print(student['name'])
print(student["age"])
```

```python
#another WAY to access dictionary elements
# Accessing using get method
print(student.get("name"))
print(student.get("last_name"))
print(student.get("last_Name" , "Not available"))  # if key is not available then it will give second argu...
```

```python
# updating the key 
student["name"] = "harry"
print(student["name"])
# del - this will delete the key along with it's vallue
del student["age"]
print(student)
```
