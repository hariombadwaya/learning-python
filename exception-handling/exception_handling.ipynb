```python
num = 10
result = num / 0
print(result)
```

Exception handling allows us to say:
"If something goes wrong, don't crash. Handle the problem gracefully."

```python
try:
    num = int(input("Enter a number: "))
    print(num)
except:
    print("Something went wrong!")

# it the user inters 10 output will be 10
# if user inter hello output will be something went wrong    
```

```python
#Catching a Specific Exception

#You generally shouldn't use a bare except when you can catch a specific exception.

try:
    num = int(input("Enter a number: "))

except ValueError:
    print("Please enter a valid number.")
```

```python
# mutiple except blocks 

try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

    print(result)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("Second number cannot be zero.")
```

```python
#Python also provides an else block.

#else runs only when the try block succeeds.

try:
    num = int(input("Enter a number: "))

except ValueError:
    print("Invalid number.")

else:
    print("You entered:", num)
```

```python
# finally is used for code that should run whether an exception happens or not.

try:
    num = 10 / 2
    print(num)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program finished.")
```

```python
#You can combine everything:

try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Execution completed.")
```

```python
# what happens after exception

try:
    print("A")
    x = 10 / 0
    print("B")

except ZeroDivisionError:
    print("C")

print("D")
```

```python
#Raising an Exception Yourself

#Python doesn't have to be the only thing that raises exceptions.

age = 15

if age < 18:
    raise ValueError("You must be at least 18.")
```

```python
try:
    age = 15

    if age < 18:
        raise ValueError("You must be at least 18.")

except ValueError as e:
    print(e)
```

```python
#Custom Exceptions

#You can even create your own exception types.

class InsufficientBalanceError(Exception):
    pass
```

```python
balance = 500
withdraw = 1000

if withdraw > balance:
    raise InsufficientBalanceError("Not enough balance.")
```

```python
# you can catch it like 

try:
    balance = 500
    withdraw = 1000

    if withdraw > balance:
        raise InsufficientBalanceError("Not enough balance.")

except InsufficientBalanceError as e:
    print(e)
```

```python
# we can catch more general exception like this 

try:
    x = int("hello")

except Exception:
    print("Something went wrong.")
```
