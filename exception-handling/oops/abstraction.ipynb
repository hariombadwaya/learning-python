Abstraction - Abstraction means hiding unnecessary implementation details and showing only the essential functionality to the user.

```python
from abc import ABC, abstractmethod   # Python provides the abc module to implement abstraction.

class Vehicle(ABC):
    def drive(self):
        print("the vehicle is moving")
    
    @abstractmethod
    def start(self):
       pass

class Car(Vehicle):
    def start(self):
        print("car engine started")

def operate_vehicle(vehicle):
    vehicle.start()

car = Car()
operate_vehicle(car)
```
