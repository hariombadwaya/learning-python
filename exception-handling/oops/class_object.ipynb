What is a Class?

A class is like a blueprint/template for creating objects

```python
class Car:
    pass
# but we haven't create acutal car yet
```

What is an Object?

An object is an actual instance of a class.

```python
class Car:
    pass


car1 = Car()
car2 = Car()

# Car
#  ↓
# Blueprint

# car1 → actual Car object
# car2 → actual Car object
```

Adding Properties

Let's say every car has:

brand
color
model

We can create them using __init__().

```python
class Car:

    def __init__(self, brand, color, model):
        self.brand = brand     #self means:   the current object.
        self.color = color
        self.model = model

        # now create an object 
car1 = Car("BMW", "Black", "X5")  # __init__() automatically when you create the object. 
print(car1.brand)  
print(car1.color)      
```

```python
car2 = Car("Audi", "White", "qa")
print(car2.brand)
print(car2.color)
```

Methods

A method is a function that belongs to an object/class.

```python
class Car:

    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def start(self):
        print("Car is starting")

    def stop(self):
        print("Car is stopping")
```

```python
# creating an object 
car1 = Car("BMW", "Black")
```

```python
# Call the methods:
car1.start()
car1.stop()
```

```python
#  Encapsulation = keeping related data and methods together and controlling access to internal data.

class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")
```

```python
account = BankAccount(1000)

account.deposit(500)
account.withdraw(200)

print(account.balance)
```

```python
# Inheritance

#  Inheritance = one class can inherit properties and methods from another class.

class Animal:

    def eat(self):
        print("Animal is eating")

    def sleep(self):
        print("Animal is sleeping")
```

```python
# dog inherits from animal
class Dog(Animal):
    pass

# therefore 

dog = Dog()   # creating object of dog class

dog.eat()
dog.sleep()
```

```python
#  Adding Dog-specific behavior
 
class Dog(Animal):

    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()
dog.sleep()
dog.bark()
```

```python


# Polymorphism

# The word sounds complicated, but the idea is simple:

# Same method name, different behavior.

class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```
