```python
# celcius to fehrenheit and fehrenheit to celcius
def convert_temperature(temp, unit):
    if unit=="c":
        return temp* 9/5 + 32   # celcius to fehrenheit
    elif unit=="f":
        return  (temp-32)* 5/9  # fehrenheit to celcius
    else:
        return None

print(convert_temperature(25,"c"))
print(convert_temperature(77,"f"))
```

Q : Password strength checker

```python
def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+/' for char in password):
        return False
    return True


print(is_strong_password("9w90iWwj/wgj"))
    
```

Q : calculate the total cost of cart items

```python
def calculate_total_cost(cart):
    total =0
    for item in cart:
        total+= item["price"]* item["quantity"]
    return total

# example of cart data
cart= [
    {"name": "apple", "price": 0.5, "quantity":4},
    {"name": "banana", "price": 0.3, "quantity":6}
]

#calling the function
total_cost = calculate_total_cost(cart)
print(total_cost)
        
```
